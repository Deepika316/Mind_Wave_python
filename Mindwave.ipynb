```python
a="deepu is a brave girl"

print(a.capitalize())      # Capitalize first letter
print(a.casefold())        # Lowercase
print(a.center(40))        # Center align
print(a.count("e"))        # Count occurrences
print(a.encode())          # Convert to bytes
print(a.endswith("girl"))  # Check ending
print(a.endswith("good"))
print(a.expandtabs())      # Expand tabs
print(a.find("is"))        # Find substring
print(a.format())          # Format string

print(a.index("is"))       # Index of substring
print(a.isalnum())         # Alphanumeric check
print(a.isalpha())         # Alphabetic check
print(a.isascii())         # ASCII check
print("123".isdecimal())   # Decimal check
print("123".isdigit())     # Digit check
print("var1".isidentifier()) # Identifier check
print(a.islower())         # Lowercase check
print("123".isnumeric())   # Numeric check
print(a.isprintable())     # Printable check
print("   ".isspace())     # Space check
print("Deepu Is A Brave Girl".istitle()) # Title check
print(a.isupper())         # Uppercase check

print("-".join(["deepu","girl"])) # Join
print(a.ljust(30))         # Left justify
print(a.lower())           # Lowercase
print("   deepu".lstrip()) # Remove left spaces

table = str.maketrans("d", "D")
print(a.translate(table))  # Translate characters

print(a.partition("is"))   # Partition
print("unhappy".removeprefix("un")) # Remove prefix
print("file.txt".removesuffix(".txt")) # Remove suffix

print(a.replace("deepu","Deepika")) # Replace
print(a.rfind("a"))        # Find from right
print(a.rindex("a"))       # Index from right
print(a.rjust(30))         # Right justify
print(a.rpartition("a"))   # Right partition
print(a.rsplit(" ",2))     # Right split
print("deepu   ".rstrip()) # Remove right spaces

print(a.split())           # Split string
print("hello\nworld".splitlines()) # Split lines

print(a.startswith("deepu")) # Check beginning
print("  deepu  ".strip())  # Remove spaces
print("DeEpU".swapcase())   # Swap case
print(a.title())            # Title case
print(a.upper())            # Uppercase
print("25".zfill(5))        # Fill with zeros
```

```python
#set operations
x={"deepu","nagamani","satyanarayana",2,4}
y={"deepu","nagamani","satyanarayana",20,10}
print(x.union(y))
print(x.intersection(y))
print(x.difference(y))
print(x.symmetric_difference(y))

z={"deepu","nagamani","satyanarayana",2,4}
print(x.issubset(z))
print(x.issuperset(y))
print(x.isdisjoint(y))
```
