```python
#lambda with map
names=["Deepu","Nagamani","Satyanarayana"]
x=map(lambda x:x.upper(),names)
print(set(x))
```

```python
names=["Deepu","Nagamani","Satyanarayana"]
x=filter(lambda x:len(x)>=6,names)
print(list(x))
```

```python
from functools import reduce
names=["Deepu","Nagamani","Satyanarayana"]
x=reduce(lambda x,y:x+y,names)
print(list(x))
```
